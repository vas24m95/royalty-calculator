```python
artist = "Drake"
revenue = 20000
royalty_rate = 0.15

royalty_due = revenue * royalty_rate

print("Artist:", artist)
print(f"Royalty due: £{royalty_due:,.2f}")
```

```python
def calculate_royalty(revenue, royalty_rate):
    return revenue * royalty_rate

royalty_due = calculate_royalty(50000, 0.70)

print(f"Royalty due: £{royalty_due:,.2f}")
```
