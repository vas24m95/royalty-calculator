```python
#import csv file

import pandas as pd
df = pd.read_csv("Spotify Streaming Performance Dataset.csv")
print(df.head())
```

```python
#check to see if all columns have been correctly imported into Google Colab
print(df.columns.tolist())
print(df.shape)
```

```python
#clean any spaces in the column names and assign clean column names back to df.columns

df.columns = df.columns.str.strip()
print(df.columns.tolist)
```

```python
# Sorts top 10 artists by total streams

top_10 = df.sort_values(
    by= 'Total Streams (in millions)',
    ascending= False).head(10)

print(top_10[["Artist Name","Total Streams (in millions)"]])
```

```python
# convert streams from millions into actual streams

top_10["Actual Streams"] = top_10["Total Streams (in millions)"] * 1000000
print(top_10[["Artist Name","Total Streams (in millions)","Actual Streams"]])
```

```python
#Estimating gross revenue, assuming Spotify pays 0.003 pence per stream

revenue_per_stream = 0.003
top_10["Estimated_Gross_Revenue"] = top_10["Actual Streams"] * revenue_per_stream
print(top_10[["Artist Name","Total Streams (in millions)","Actual Streams","Estimated_Gross_Revenue"]])
```

```python
#Estimating artist payout based on contractual share of 20%

artist_royalty_rate = 0.20
top_10["Estimated Artist Payout (£)"] = top_10["Estimated_Gross_Revenue"] * artist_royalty_rate
print(top_10[["Artist Name","Total Streams (in millions)","Actual Streams","Estimated_Gross_Revenue","Estimated Artist Payout (£)"]])
```

```python
#Creating a clean report for our top 10 artists

report = top_10[
    [
        "Artist Name",
        "Total Streams (in millions)",
        "Estimated_Gross_Revenue",
        "Estimated Artist Payout (£)"
    ]
    ].copy()


#Renaming columns for final report

report = report.rename(columns={
    "Total Streams (in millions)": "Streams (millions)",
    "Estimated_Gross_Revenue": "Estimated Gross Revenue (£)",

})

#Displaying the report with formatted numbers/values

display(
    report.style.format({
        "Streams (millions)": "{:,.2f}".format,
        "Estimated Gross Revenue (£)": "£{:,.2f}".format,
        "Estimated Artist Payout (£)": "£{:,.2f}".format,
    })
)
```

```python
#Define the assumed artist royalty rates

royalty_rates = [0.10, 0.20, 0.30]

#calculating the payouts in each of the assumed rates
for rate in royalty_rates:
  report[f"payout at {rate:.0%} (£)"] = (
      report["Estimated Gross Revenue (£)"] * rate
  )




#Display the full report with the assumed rates
display(
    report.style.format({
        "Estimated Gross Revenue (£)": "£{:,.2f}".format,
        "Estimated Artist Payout (£)": "£{:,.2f}".format,
        "payout at 10% (£)": "£{:,.2f}".format,
        "payout at 20% (£)": "£{:,.2f}".format,
        "payout at 30% (£)": "£{:,.2f}".format
    })

)
```

```python
#Exporting the report as a CSV file

report.to_csv("royalty_calculator_v2_report.csv", index=False)

print("Report exported succesfully")
```
